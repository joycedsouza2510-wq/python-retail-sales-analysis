```python
# Calculate your first business KPI using Pandas(Sales,total revenue,profit)

df["Sales"] = df["Quantity"] * df["Unit_Price"]

print(df[["Product", "Quantity", "Unit_Price", "Sales"]])

total_revenue = df["Sales"].sum()

print(f"Total revenue: €{total_revenue:,.2f}")

df["Profit"] = (
    df["Quantity"] *
    (df["Unit_Price"] - df["Cost_Per_Unit"])
)

total_profit = df["Profit"].sum()

print(f"Total profit: €{total_profit:,.2f}")

```

```python
# Data cleaning with Pandas
df_dirty = df.copy()

# Introduce practice data-quality problems
df_dirty.loc[1, "Region"] = None
df_dirty.loc[3, "Product"] = " signal boosters "
df_dirty.loc[4, "Unit_Price"] = -820

# Add a duplicate row
df_dirty = pd.concat(
    [df_dirty, df_dirty.iloc[[0]]],
    ignore_index=True
)

print(df_dirty.shape)


# Check missing values
print(df_dirty.isnull().sum())

# Check duplicates
print("Duplicates:", df_dirty.duplicated().sum())

# Check negative prices
print(df_dirty[df_dirty["Unit_Price"] < 0])


# Remove exact duplicate rows
df_clean = df_dirty.drop_duplicates().copy()

# Standardise product names
df_clean["Product"] = (
    df_clean["Product"].str.strip().str.title()
)

# Fill missing regions
df_clean["Region"] = df_clean["Region"].fillna("Unknown")

# Remove invalid negative prices for this exercise
df_clean = df_clean[df_clean["Unit_Price"] >= 0].copy()

print("Cleaned rows:", len(df_clean))
print("Missing values:\n", df_clean.isnull().sum())
print("Duplicates:", df_clean.duplicated().sum())
```

```python
import pandas as pd

# Ensure the dataset is loaded before creating a copy
try:
    df_eda = df.copy()
except NameError:
    print("df was not found in memory. Attempting to load 'retail_sales.csv'...")
    try:
        df = pd.read_csv("retail_sales.csv")
        df_eda = df.copy()
    except FileNotFoundError:
        print("CSV file not found on disk. Re-creating the sample dataframe...")
        data = {
            "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
            "Product": ["Television", "Soundbars", "HDMI cables", "signal boosters",
                        "Television", "Soundbars", "signal boosters", "HDMI cables",
                        "Television", "Soundbars"],
            "Category": ["Electronics", "Accessories", "Accessories",
                         "Accessories", "Electronics", "Accessories",
                         "Accessories", "Accessories", "Electronics",
                         "Accessories"],
            "Quantity": [2, 5, 3, 1, 1, 10, 2, 4, 3, 6],
            "Unit_Price": [820, 120, 75, 220, 820, 120, 220, 75, 820, 120],
            "Cost_Per_Unit": [600, 12, 20, 150, 600, 12, 150, 20, 600, 12],
            "Region": ["Dublin", "Cork", "Galway", "Dublin", "Cork",
                       "Dublin", "Galway", "Cork", "Dublin", "Galway"]
        }
        df = pd.DataFrame(data)
        df.to_csv("retail_sales.csv", index=False)
        df_eda = df.copy()

# Calculate sales revenue for each transaction
df_eda["Sales"] = (
    df_eda["Quantity"] * df_eda["Unit_Price"]
)

# Calculate profit for each transaction
df_eda["Profit"] = (
    df_eda["Quantity"] *
    (df_eda["Unit_Price"] - df_eda["Cost_Per_Unit"])
)

print(df_eda.head())
```

```python
# Calculate business KPIs
total_revenue = df_eda["Sales"].sum()
total_profit = df_eda["Profit"].sum()
total_units = df_eda["Quantity"].sum()

profit_margin = (total_profit / total_revenue) * 100

print(f"Total revenue: €{total_revenue:,.2f}")
print(f"Total profit: €{total_profit:,.2f}")
print(f"Total units sold: {total_units}")
print(f"Profit margin: {profit_margin:.2f}%")
```

```python
# Find which product generates the most revenue

product_sales = (
    df_eda.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(product_sales)
```

```python
# Compare regions

region_sales = (
    df_eda.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(region_sales)
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Load the original dataset
df = pd.read_csv("retail_sales.csv")

# Check the dataset
print(df.shape)
print(df.head())
```

```python

# Calculate revenue and gross profit
df["Sales"] = df["Quantity"] * df["Unit_Price"]

df["Profit"] = (
    df["Quantity"] *
    (df["Unit_Price"] - df["Cost_Per_Unit"])
)

# Calculate KPIs
total_revenue = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_units = df["Quantity"].sum()
total_orders = df["Order_ID"].nunique()

profit_margin = total_profit / total_revenue * 100
average_order_value = total_revenue / total_orders

print(f"Revenue: €{total_revenue:,.2f}")
print(f"Gross profit: €{total_profit:,.2f}")
print(f"Units sold: {total_units}")
print(f"Orders: {total_orders}")
print(f"Gross profit margin: {profit_margin:.2f}%")
print(f"Average order value: €{average_order_value:.2f}")
```

```python

# Sales and profit by product
product_analysis = (
    df.groupby("Product")[["Sales", "Profit", "Quantity"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print(product_analysis)

# Sales by region
region_analysis = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(region_analysis)
```

```python
# Chart 1: Revenue by product

product_analysis["Sales"].plot(
    kind="bar",
    figsize=(8, 5),
    color="steelblue"
)

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (€)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("product_revenue.png", dpi=150)
plt.show()
```

```python
# Chart 2: Revenue by region

region_analysis.plot(
    kind="bar",
    figsize=(8, 5),
    color="seagreen"
)

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (€)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("regional_revenue.png", dpi=150)
plt.show()
```

```python
# Chart 3: Gross profit by product

product_analysis["Profit"].plot(
    kind="bar",
    figsize=(8, 5),
    color="orange"
)

plt.title("Gross Profit by Product")
plt.xlabel("Product")
plt.ylabel("Gross Profit (€)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("product_profit.png", dpi=150)
plt.show()
```
