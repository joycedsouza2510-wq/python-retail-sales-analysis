```python
# Create a sample retail CSV file
import pandas as pd

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],
    "Product": ["Television", "Soundbars", "HDMI cables", "signal boosters",
                "Television", "Soundbars", "signal boosters", "HDMI cables",
                "Television", "Soundbars"],
    "Category": ["Electronics", "Accessories", "Accessories",
                 "Accessories", "Electronics", "Accessories",
                 "Accessories", "Accessories", "Electronics",
                 "Accessories"],

    "Quantity": [2, 5, 3, 1, 1, 10, 2, 4, 3, 6],
    "Unit_Price": [820, 120, 75, 220, 820, 120, 220, 75, 820, 120],
    "Cost_Per_Unit": [600, 12, 20, 150, 600, 12, 150, 20, 600, 12],
    "Region": ["Dublin", "Cork", "Galway", "Dublin", "Cork",
               "Dublin", "Galway", "Cork", "Dublin", "Galway"]
}

df = pd.DataFrame(data)

df.to_csv("retail_sales.csv", index=False)

print("CSV file created successfully!")
```

```python
# Load the CSV file

df = pd.read_csv("retail_sales.csv")

print(df)
```

```python
# Explore the dataset

print(df.head())
print(df.shape)

print(df.columns)

df.info()


```

```python
# Calculate your first business KPI using Pandas(Sales,total revenue,profit)

df["Sales"] = df["Quantity"] * df["Unit_Price"]

print(df[["Product", "Quantity", "Unit_Price", "Sales"]])

total_revenue = df["Sales"].sum()

print(f"Total revenue: €{total_revenue:,.2f}")

df["Profit"] = (
    df["Quantity"] *
    (df["Unit_Price"] - df["Cost_Per_Unit"])
)

total_profit = df["Profit"].sum()

print(f"Total profit: €{total_profit:,.2f}")

```

```python
# Data cleaning with Pandas
df_dirty = df.copy()

# Introduce practice data-quality problems
df_dirty.loc[1, "Region"] = None
df_dirty.loc[3, "Product"] = " signal boosters "
df_dirty.loc[4, "Unit_Price"] = -820

# Add a duplicate row
df_dirty = pd.concat(
    [df_dirty, df_dirty.iloc[[0]]],
    ignore_index=True
)

print(df_dirty.shape)


# Check missing values
print(df_dirty.isnull().sum())

# Check duplicates
print("Duplicates:", df_dirty.duplicated().sum())

# Check negative prices
print(df_dirty[df_dirty["Unit_Price"] < 0])


# Remove exact duplicate rows
df_clean = df_dirty.drop_duplicates().copy()

# Standardise product names
df_clean["Product"] = (
    df_clean["Product"].str.strip().str.title()
)

# Fill missing regions
df_clean["Region"] = df_clean["Region"].fillna("Unknown")

# Remove invalid negative prices for this exercise
df_clean = df_clean[df_clean["Unit_Price"] >= 0].copy()

print("Cleaned rows:", len(df_clean))
print("Missing values:\n", df_clean.isnull().sum())
print("Duplicates:", df_clean.duplicated().sum())
```
